## Part 2: Setting Up Azure Resources

Before creating the MCP server, we need:
1. **Azure API Center** - MCP server registry (with portal access configured)
2. **Standard Logic App** - Execution runtime for connector actions

Both resources must be in the **same Azure subscription**.

````python
# Configuration for Azure resources
import os
from dotenv import load_dotenv

load_dotenv()

# Resource configuration - customize these for your environment
RESOURCE_GROUP = os.getenv("AZURE_RESOURCE_GROUP", "rg-mcp-servers")
LOCATION = os.getenv("AZURE_LOCATION", "eastus")
API_CENTER_NAME = os.getenv("API_CENTER_NAME", "mcp-api-center")  # Must be globally unique
LOGIC_APP_NAME = os.getenv("LOGIC_APP_NAME", "travel-assistant-mcp")
STORAGE_ACCOUNT = os.getenv("STORAGE_ACCOUNT", "mcplogicappstorage")  # lowercase, no hyphens

print("=" * 60)
print("Azure Resource Configuration for Logic Apps MCP Server")
print("=" * 60)
print(f"\n  Resource Group:   {RESOURCE_GROUP}")
print(f"  Location:         {LOCATION}")
print(f"  API Center:       {API_CENTER_NAME}")
print(f"  Logic App:        {LOGIC_APP_NAME}")
print(f"  Storage Account:  {STORAGE_ACCOUNT}")
print("\n" + "=" * 60)
print("\nIMPORTANT: API Center and Logic App must be in the SAME subscription!")
print("=" * 60)
````

````python
# Create deployment script for Azure resources
import stat

deployment_script = f'''#!/bin/bash
set -e

echo "================================================================"
echo "Creating Azure Resources for Logic Apps MCP Server"
echo "================================================================"

# Variables
RESOURCE_GROUP="{RESOURCE_GROUP}"
LOCATION="{LOCATION}"
API_CENTER_NAME="{API_CENTER_NAME}"
LOGIC_APP_NAME="{LOGIC_APP_NAME}"
STORAGE_ACCOUNT="{STORAGE_ACCOUNT}"

# 1. Create Resource Group (if not exists)
echo ""
echo "Step 1: Creating resource group..."
az group create \\
  --name $RESOURCE_GROUP \\
  --location $LOCATION \\
  --output table

# 2. Create Azure API Center
echo ""
echo "Step 2: Creating Azure API Center..."
az apic create \\
  --name $API_CENTER_NAME \\
  --resource-group $RESOURCE_GROUP \\
  --location $LOCATION \\
  --output table

# 3. Create Storage Account for Logic App
echo ""
echo "Step 3: Creating storage account for Logic App..."
az storage account create \\
  --name $STORAGE_ACCOUNT \\
  --resource-group $RESOURCE_GROUP \\
  --location $LOCATION \\
  --sku Standard_LRS \\
  --output table

# 4. Create Standard Logic App (Workflow Service Plan)
echo ""
echo "Step 4: Creating Standard Logic App..."
az logicapp create \\
  --name $LOGIC_APP_NAME \\
  --resource-group $RESOURCE_GROUP \\
  --storage-account $STORAGE_ACCOUNT \\
  --output table

# 5. Get API Center Portal URL
echo ""
echo "Step 5: Getting API Center Portal URL..."
PORTAL_URL=$(az apic show \\
  --name $API_CENTER_NAME \\
  --resource-group $RESOURCE_GROUP \\
  --query "properties.portalUrl" -o tsv 2>/dev/null || echo "Portal not ready")

echo ""
echo "================================================================"
echo "Resources Created Successfully!"
echo "================================================================"
echo ""
echo "API Center Portal: https://$API_CENTER_NAME.$LOCATION.azure-apicenter.ms"
echo ""
echo "Next Steps:"
echo "1. Open Azure Portal: https://portal.azure.com"
echo "2. Navigate to API Center: $API_CENTER_NAME"
echo "3. Go to: Discovery > MCP (Preview)"
echo "4. Click 'Register' on the Azure Logic Apps tile"
echo "5. Select Logic App: $LOGIC_APP_NAME"
echo "6. Add connector actions as MCP tools"
echo ""
echo "================================================================"
'''

# Save deployment script
with open('deploy_logic_app_resources.sh', 'w') as f:
    f.write(deployment_script)

os.chmod('deploy_logic_app_resources.sh', os.stat('deploy_logic_app_resources.sh').st_mode | stat.S_IEXEC)

print("✅ Created: deploy_logic_app_resources.sh")
print("\nTo create Azure resources, run:")
print("  ./deploy_logic_app_resources.sh")
print("\nOr run the commands manually in the Azure Portal Cloud Shell.")
````

### How OAuth 2.0 Authentication Works

2. Include the token in the `Authorization: Bearer <token>` header

````python
# Debug: Inspect the token claims
import base64
import json

def decode_jwt(token):
    """Decode and display JWT claims (header and payload)"""
    parts = token.split('.')
    if len(parts) != 3:
        print("Invalid JWT format")
        return
    
    # Decode header and payload (add padding if needed)
    for i, name in enumerate(["Header", "Payload"]):
        part = parts[i]
        # Add padding
        padding = 4 - len(part) % 4
        if padding != 4:
            part += '=' * padding
        try:
            decoded = base64.urlsafe_b64decode(part)
            data = json.loads(decoded)
            print(f"\n{name}:")
            print(json.dumps(data, indent=2))
        except Exception as e:
            print(f"Error decoding {name}: {e}")

# Get a fresh token and inspect it
from azure.identity import InteractiveBrowserCredential

credential = InteractiveBrowserCredential(
    tenant_id="3ce22404-c8d7-4655-9180-3d433c40edaf"
)
token = credential.get_token("api://6344333e-84b7-4a67-876b-47db0a5c86d5/user_impersonation")

print("=" * 70)
print("JWT Token Claims Analysis")
print("=" * 70)
decode_jwt(token.token)

print("\n" + "=" * 70)
print("Token Audience Check")
print("=" * 70)
# The 'aud' claim should match what Easy Auth expects
print("\nExpected audience: api://6344333e-84b7-4a67-876b-47db0a5c86d5/")
````

## Part 8: Programmatic Access Considerations

### Logic Apps MCP vs FastMCP: Authentication Differences

| Feature | Logic Apps MCP (API Center) | FastMCP |
|---------|----------------------------|---------|
| **Authentication** | Interactive OAuth via VS Code | API keys, custom headers, or none |
| **Best for** | End-user tools in VS Code/Copilot | Programmatic agent access |
| **Setup** | Azure Portal UI (no-code) | Python code |
| **Connectors** | 1,400+ enterprise connectors | Custom Python functions |

### When to Use Each Approach

**Use Logic Apps MCP when:**
- Building tools for human users in VS Code
- Need enterprise connectors (SAP, Salesforce, O365)
- Want no-code tool creation
- OAuth/Entra ID auth is acceptable

**Use FastMCP when:**
- Building automated agent pipelines
- Need programmatic (non-interactive) access
- Want custom authentication (API keys)
- Need maximum flexibility in tool logic

````python
# Alternative: Using FastMCP for Programmatic Access
# If you need programmatic (non-interactive) access to Logic Apps connectors,
# consider wrapping them in a FastMCP server with custom authentication.

fastmcp_alternative = '''
"""
Alternative Approach: Wrap Logic Apps in FastMCP for Programmatic Access

Instead of using the API Center MCP server directly (which requires 
interactive OAuth), you can:

1. Create an Azure Function that calls Logic Apps HTTP triggers
2. Wrap that function in a FastMCP server
3. Use API key or custom auth for your agent

Example architecture:
┌─────────────────┐     ┌─────────────────┐     ┌─────────────────┐
│  Agent          │     │  FastMCP        │     │  Azure          │
│  Framework      │────▶│  Server         │────▶│  Function       │
│                 │ MCP │  (API Key Auth) │ HTTP│                 │
└─────────────────┘     └─────────────────┘     └─────────────────┘
                                                        │
                                                        ▼
                                                ┌─────────────────┐
                                                │  Logic Apps     │
                                                │  Workflow       │
                                                │  (HTTP Trigger) │
                                                └─────────────────┘
"""

# FastMCP server that wraps Logic Apps
from mcp.server.fastmcp import FastMCP

mcp = FastMCP("logic-apps-wrapper")

@mcp.tool(description="Send email via Logic Apps workflow")
async def send_email(to: str, subject: str, body: str) -> str:
    """Calls a Logic Apps workflow via HTTP trigger"""
    import aiohttp
    
    # Logic Apps HTTP trigger URL (with SAS token for auth)
    workflow_url = os.getenv("LOGIC_APP_SEND_EMAIL_URL")
    
    async with aiohttp.ClientSession() as session:
        async with session.post(
            workflow_url,
            json={"to": to, "subject": subject, "body": body}
        ) as resp:
            if resp.status == 200:
                return f"Email sent to {to}"
            else:
                return f"Failed: {await resp.text()}"

# This server can then be used with HostedMCPTool or MCPStreamableHTTPTool
# with API key authentication for programmatic access
'''

print("Alternative Approach: FastMCP Wrapper for Programmatic Access")
print("=" * 70)
print(fastmcp_alternative)
print("=" * 70)
print("""
Summary:
- Logic Apps MCP (via API Center) is designed for VS Code interactive use
- For programmatic agent access, consider:
  1. FastMCP server wrapping Logic Apps HTTP triggers
  2. Direct Azure Function calls from your agent
  3. Using the connector SDKs directly (e.g., Microsoft Graph for O365)
""")
````
